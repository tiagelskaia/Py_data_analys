```python
 #   Column            Dtype       оптимальный тип
---  ------            --------------  ----- 
 #   price                 int64      int32
 #   area                  int64      int16
 #   bedrooms              int64      int8
 #   bathrooms             int64      int8
 #   stories               int64      int8
 #   mainroad              object     category
 #   guestroom             object     category
 #   basement              object     category
 #   hotwaterheating       object     category
 #  airconditioning       object      category
 #  parking               int64       int8
 #  prefarea              object      category
 #  furnishingstatus      object      category
```

```python
#поменяйтие типы данных и сравните потребление памяти
```

```python
df['price'] = df['price'].astype(np.int32)
df['area'] = df['area'].astype(np.int16)
df['bedrooms'] = df['bedrooms'].astype(np.int8)
df['bathrooms'] = df['bathrooms'].astype(np.int8)
df['stories'] = df['stories'].astype(np.int8)
df['parking'] = df['parking'].astype(np.int8)
```

```python
df['mainroad'] = df['mainroad'].astype("category")
df['guestroom'] = df['guestroom'].astype("category")
df['basement'] = df['basement'].astype("category")
df['hotwaterheating'] = df['hotwaterheating'].astype("category")
df['airconditioning'] = df['airconditioning'].astype("category")
df['prefarea'] = df['prefarea'].astype("category")
df['furnishingstatus'] = df['furnishingstatus'].astype("category")
```

```python
df.info()
```

```python
#dtypes: int64(6), object(7), memory usage: 55.5+ KB
#dtypes: category(7), int16(1), int32(1), int8(4), memory usage: 10.0 KB
# памяти требуется 10кБ вместо 55,5кБ, что примерно в 5,5 раз меньше
```
