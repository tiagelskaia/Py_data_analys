```python
import requests
import pandas as pd
```

```python
api_base_url = 'https://www.anapioficeandfire.com/api'
```

```python
#получить документацию обо всех книгах.
books_params = {'page': 1, 'pageSize': 50}
books_list = requests.get(f'{api_base_url}/books', params = books_params).json()
len(books_list)
```

```python
#сохранить информацию о книгах в датафрейм
books_df = pd.DataFrame(books_list)
books_df.info()
```

```python
#получить информацию обо всех домах Вестероса
houses_params = {'page': 0, 'pageSize': 50}
houses_list_total = []
for n in range(1, 10):
    houses_params['page'] = n
    houses_list = requests.get(f'{api_base_url}/houses', params = houses_params).json()
    houses_list_total.extend(houses_list)
len(houses_list_total)
```

```python
#сохранить информацию о домах Вестероса в датафрейм
houses_df = pd.DataFrame(houses_list_total)
houses_df.info()
```

```python
#получить информацию обо всех домах Вестероса, у которых есть девиз, и сохранить в датафрейм - первый вариант
houses_df_words = houses_df[houses_df['words'].apply(lambda x: len(x) > 0)]
houses_df_words.info()
```

```python
#другой вариант, через запрос - девиз через параметр hasWords (опечатка в задании)
houses_params = {'page': 0, 'pageSize': 50, 'hasWords': True}
houses_list_words_total = []
for n in range(1, 3):
    houses_params['page'] = n
    houses_list_words = requests.get(f'{api_base_url}/houses', params = houses_params).json()
    houses_list_words_total.extend(houses_list_words)
len(houses_list_words_total)
```

```python
houses_words_df = pd.DataFrame(houses_list_words_total)
houses_words_df.info()
```
